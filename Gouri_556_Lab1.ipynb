```python


!pip install openpyxl -q


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

uploaded = files.upload()


file_name = "boston_housing_clean.xlsx"

df = pd.read_excel(file_name)

# Remove unwanted unnamed columns if present
df = df.loc[
    :,
    ~df.columns.astype(str).str.startswith("Unnamed")
]

print("\nDataset loaded successfully!")

print("\n================================================")
print("FIRST 5 ROWS")
print("================================================")

print(df.head())

print("\n================================================")
print("DATASET DIMENSION")
print("================================================")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


print("\n================================================")
print("ATTRIBUTE NAMES")
print("================================================")

print(df.columns.tolist())



print("\n================================================")
print("DATA TYPES")
print("================================================")

print(
    df.dtypes.to_frame(
        name="Data Type"
    )
)


print("\n================================================")
print("DATASET INFORMATION")
print("================================================")

df.info()


print("\n================================================")
print("MISSING VALUES")
print("================================================")

missing_values = df.isnull().sum()

print(
    missing_values.to_frame(
        name="Missing Values"
    )
)

if missing_values.sum() == 0:

    print("No missing values found.")

else:

    print("Missing values are present.")


print("\n================================================")
print("DUPLICATE RECORDS")
print("================================================")

duplicate_count = df.duplicated().sum()

print(
    "Number of duplicate records:",
    duplicate_count
)

if duplicate_count == 0:

    print("No duplicate records found.")

else:

    print("Duplicate records are present.")


print("\n================================================")
print("BASIC STATISTICAL INFORMATION")
print("================================================")

print(
    df.describe()
)


print("\n================================================")
print("TARGET VARIABLE")
print("================================================")

target_column = None

for column in df.columns:

    if str(column).strip().upper() == "MEDV":

        target_column = column
        break


if target_column is None:

    print("ERROR: MEDV column was not found.")

    print("Available columns:")

    print(
        df.columns.tolist()
    )

    raise ValueError(
        "MEDV column not found."
    )


print(
    "Target variable:",
    target_column
)


#handling ,missing values

numeric_columns = df.select_dtypes(
    include=np.number
).columns


df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)




print("\n================================================")
print("EXPLORATORY DATA ANALYSIS")
print("================================================")


# Distribution of MEDV


plt.figure(
    figsize=(8, 5)
)

sns.histplot(
    df[target_column],
    kde=True
)

plt.title(
    "Distribution of MEDV"
)

plt.xlabel(
    "MEDV"
)

plt.ylabel(
    "Frequency"
)

plt.show()


# 16. SELECTED FEATURES VS MEDV


selected_features = [
    "CRIM",
    "RM",
    "LSTAT",
    "PTRATIO",
    "TAX"
]


# Keep only features that actually exist
selected_features = [
    feature
    for feature in selected_features
    if feature in df.columns
]


for feature in selected_features:

    plt.figure(
        figsize=(7, 5)
    )

    sns.scatterplot(
        data=df,
        x=feature,
        y=target_column
    )

    plt.title(
        feature + " vs MEDV"
    )

    plt.xlabel(
        feature
    )

    plt.ylabel(
        "MEDV"
    )

    plt.show()



print("\n================================================")
print("CORRELATION WITH MEDV")
print("================================================")

correlation = df.corr(
    numeric_only=True
)

medv_correlation = (
    correlation[target_column]
    .sort_values(
        ascending=False
    )
)

print(
    medv_correlation.to_frame(
        name="Correlation with MEDV"
    )
)

plt.figure(
    figsize=(12, 9)
)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(
    "Correlation Matrix / Heatmap"
)

plt.tight_layout()

plt.show()



print("\n================================================")
print("STRONG RELATIONSHIPS WITH MEDV")
print("================================================")

absolute_correlation = (
    correlation[target_column]
    .drop(target_column)
    .abs()
    .sort_values(
        ascending=False
    )
)

print(
    absolute_correlation.to_frame(
        name="Absolute Correlation"
    )
)




print("\n================================================")
print("DATA PREPARATION")
print("================================================")


# Predictor variables
X = df.drop(
    target_column,
    axis=1
)


# Target variable
y = df[target_column]


# Keep only numerical predictor variables
X = X.select_dtypes(
    include=np.number
)


# Handle missing values
X = X.fillna(
    X.median()
)


print(
    "Predictor variables:"
)

print(
    X.columns.tolist()
)

print(
    "\nShape of X:",
    X.shape
)

print(
    "Shape of y:",
    y.shape
)




print("\n================================================")
print("TRAIN-TEST SPLIT")
print("================================================")


X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42
)


print(
    "Training samples:",
    X_train.shape[0]
)

print(
    "Testing samples:",
    X_test.shape[0]
)




model = LinearRegression()



model.fit(
    X_train,
    y_train
)


print("\n================================================")
print("MODEL TRAINING")
print("================================================")

print(
    "Multiple Linear Regression model trained successfully!"
)


y_train_pred = model.predict(
    X_train
)

y_test_pred = model.predict(
    X_test
)


print(
    "Predictions generated successfully!"
)



print("\n================================================")
print("MODEL INTERCEPT")
print("================================================")

print(
    "Intercept:",
    model.intercept_
)




print("\n================================================")
print("REGRESSION COEFFICIENTS")
print("================================================")


coefficient_df = pd.DataFrame({

    "Feature": X.columns,

    "Coefficient": model.coef_

})


coefficient_df[
    "Absolute Coefficient"
] = coefficient_df[
    "Coefficient"
].abs()


coefficient_df = coefficient_df.sort_values(

    by="Absolute Coefficient",

    ascending=False

)


print(
    coefficient_df
)



# MAE
mae = mean_absolute_error(
    y_test,
    y_test_pred
)


# MSE
mse = mean_squared_error(
    y_test,
    y_test_pred
)


# RMSE
rmse = np.sqrt(
    mse
)


# R2
r2 = r2_score(
    y_test,
    y_test_pred
)


# ================================================================
# 28. ADJUSTED R2
# ================================================================

n = X_test.shape[0]

p = X_test.shape[1]


adjusted_r2 = 1 - (

    ((1 - r2) * (n - 1))

    /

    (n - p - 1)

)



print("\n================================================")
print("MODEL EVALUATION")
print("================================================")

print(
    "Mean Absolute Error (MAE):",
    round(mae, 4)
)

print(
    "Mean Squared Error (MSE):",
    round(mse, 4)
)

print(
    "Root Mean Squared Error (RMSE):",
    round(rmse, 4)
)

print(
    "R² Score:",
    round(r2, 4)
)

print(
    "Adjusted R²:",
    round(adjusted_r2, 4)
)



train_r2 = r2_score(
    y_train,
    y_train_pred
)

test_r2 = r2_score(
    y_test,
    y_test_pred
)


train_rmse = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred
    )
)


test_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred
    )
)


print("\n================================================")
print("TRAINING VS TESTING PERFORMANCE")
print("================================================")

print(
    "Training R²:",
    round(train_r2, 4)
)

print(
    "Testing R²:",
    round(test_r2, 4)
)

print(
    "Training RMSE:",
    round(train_rmse, 4)
)

print(
    "Testing RMSE:",
    round(test_rmse, 4)
)



print("\n================================================")
print("UNDERFITTING / OVERFITTING ANALYSIS")
print("================================================")


difference = train_r2 - test_r2


print(
    "Difference between Training and Testing R²:",
    round(difference, 4)
)


if difference > 0.10:

    print(
        "The model shows some evidence of overfitting."
    )

elif train_r2 < 0.50 and test_r2 < 0.50:

    print(
        "The model may show evidence of underfitting."
    )

else:

    print(
        "There is no strong evidence of severe overfitting."
    )


print("\n================================================")
print("ACTUAL VS PREDICTED VALUES")
print("================================================")


comparison_df = pd.DataFrame({

    "Actual MEDV": y_test.values,

    "Predicted MEDV": y_test_pred

})


print(
    comparison_df.head(20)
)



plt.figure(
    figsize=(8, 6)
)


plt.scatter(

    y_test,

    y_test_pred,

    alpha=0.7

)


minimum = min(

    y_test.min(),

    y_test_pred.min()

)


maximum = max(

    y_test.max(),

    y_test_pred.max()

)


# Perfect prediction line

plt.plot(

    [minimum, maximum],

    [minimum, maximum]

)


plt.xlabel(
    "Actual MEDV"
)

plt.ylabel(
    "Predicted MEDV"
)

plt.title(
    "Actual vs Predicted MEDV"
)

plt.tight_layout()

plt.show()


# RESIDUAL CALCULATION


residuals = (
    y_test - y_test_pred
)



# RESIDUAL PLOT


plt.figure(
    figsize=(8, 6)
)


plt.scatter(

    y_test_pred,

    residuals,

    alpha=0.7

)


plt.axhline(
    y=0
)


plt.xlabel(
    "Predicted MEDV"
)

plt.ylabel(
    "Residual"
)

plt.title(
    "Residual Plot"
)

plt.tight_layout()

plt.show()


# RESIDUAL DISTRIBUTION
plt.figure(
    figsize=(8, 5)
)


sns.histplot(

    residuals,

    kde=True

)


plt.xlabel(
    "Residual"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "Distribution of Residuals"
)

plt.tight_layout()

plt.show()



# LARGEST POSITIVE COEFFICIENTS



print("LARGEST POSITIVE COEFFICIENTS")



positive_coefficients = coefficient_df.sort_values(

    by="Coefficient",

    ascending=False

)


print(

    positive_coefficients[
        [
            "Feature",
            "Coefficient"
        ]
    ].head()

)



# LARGEST NEGATIVE COEFFICIENTS


print("LARGEST NEGATIVE COEFFICIENTS")



negative_coefficients = coefficient_df.sort_values(

    by="Coefficient",

    ascending=True

)


print(

    negative_coefficients[
        [
            "Feature",
            "Coefficient"
        ]
    ].head()

)




print("\n================================================")
print("FEATURES WITH RELATIVELY LARGE COEFFICIENTS")
print("================================================")


print(

    coefficient_df[
        [
            "Feature",
            "Coefficient",
            "Absolute Coefficient"
        ]
    ].head(10)

)


print("\n================================================")
print("MODEL INTERPRETATION")
print("================================================")


print(
    "R² Score:",
    round(r2, 4)
)




# IDENTIFY LARGEST POSITIVE AND NEGATIVE FEATURES


largest_positive = coefficient_df.loc[
    coefficient_df["Coefficient"].idxmax()
]


largest_negative = coefficient_df.loc[
    coefficient_df["Coefficient"].idxmin()
]


print(
    "\nFeature with largest positive coefficient:"
)

print(
    largest_positive["Feature"],
    "=",
    round(
        largest_positive["Coefficient"],
        4
    )
)


print(
    "\nFeature with largest negative coefficient:"
)

print(
    largest_negative["Feature"],
    "=",
    round(
        largest_negative["Coefficient"],
        4
    )
)



print("\n================================================")
print("NEW HOUSE PREDICTION")
print("================================================")


print(
    "Using the mean value of each feature as an example input."
)


new_house = X.mean().to_frame().T


new_prediction = model.predict(
    new_house
)


print(
    "Predicted MEDV:",
    round(
        new_prediction[0],
        2
    )
)


print("\n================================================")
print("FINAL MODEL SUMMARY")
print("================================================")


summary_df = pd.DataFrame({

    "Metric": [

        "MAE",
        "MSE",
        "RMSE",
        "R² Score",
        "Adjusted R²",
        "Training R²",
        "Testing R²"

    ],

    "Value": [

        mae,
        mse,
        rmse,
        r2,
        adjusted_r2,
        train_r2,
        test_r2

    ]

})


print(
    summary_df
)


```
